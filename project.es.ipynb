#### Ejercicio 01. Análisis de datos básico con estructuras nativas de python.
Crea una lista con las calificaciones de 5 estudiantes: [8.5, 9.2, 7.8, 8.9, 10].

- Calcula el promedio de las calificaciones.

```python
calif_estud = [8.5, 9.2, 7.8, 8.9, 10]
print(round(sum(calif_estud)/len(calif_estud), 2))
```

- Encuentra la calificación más alta y la más baja.

```python
calif_estud = [8.5, 9.2, 7.8, 8.9, 10]
calif_max = calif_estud[0]
calif_min = calif_estud[0]

for x in calif_estud:
    if x > calif_max:
        calif_max = x
    if x < calif_min:
        calif_min = x

print(f"La calificación más alta es {calif_max} y la más baja sería {calif_min}")
```

#### Ejercicio 01. ¿Cuál es la casa más cara en todo el dataset? (★☆☆)

Imprime la dirección y el precio de la casa seleccionada. Por ejemplo:

`La casa con dirección en Calle del Prado, Nº20 es la más cara y su precio es de 5000000 USD`

```python
casa_precio_maximo = ds.loc[ds['price'] == ds['price'].max()]

print(f"La casa con dirección en {ds.loc[casa_precio_maximo.index[0], 'address']}, Nº {ds.loc[ds['price'] == ds['price'].max()].index[0]} es la más cara y su precio es de {ds['price'].max()} USD")
```

#### Ejercicio 02. ¿Cuál es la casa más barata del dataset? (★☆☆)

Este ejercicio es similar al ejercicio anterior, solo que ahora buscamos la casa con el precio más bajo. Recuerda imprimir la dirección y el precio de la casa seleccionada. Por ejemplo:

`La casa con dirección en Calle Alcalá, Nº58 es la más barata y su precio es de 12000 USD`

```python
casa_mas_barata = ds.loc[ds["price"] == ds.loc[ds['price'] > 0, "price"].min()]

print(f"La casa con dirección en {ds.loc[casa_mas_barata.index[0], 'address']}, Nº {casa_mas_barata.index[0]} es la más barata y su precio es de {ds.loc[casa_mas_barata.index[0], 'price']} USD")

```

#### Ejercicio 03. ¿Cuál es la casa más grande y la más pequeña del dataset? (★☆☆)

Imprime la dirección y el área de las casas seleccionadas. Por ejemplo:

`La casa más grande está ubicada en Calle Gran Vía, Nº38 y su superficie es de 5000 metros`

`La casa más pequeña está ubicada en Calle Mayor, Nº12 y su superficie es de 200 metros`

Este ejercicio es similar al anterior, pero buscamos las casas más grandes y pequeñas en base a su superficie.

```python
casa_mas_grande = ds.loc[ds['surface'] == ds['surface'].max()]
casa_mas_pequenha = ds.loc[ds['surface'] == ds['surface'].min()]

print(f"La casa más grande está ubicada en {ds.loc[casa_mas_grande.index[0], 'address']} y su superficie es de {ds.loc[casa_mas_grande.index[0], 'surface']}")
print(f"La casa más pequeña está ubicada en {ds.loc[casa_mas_pequenha.index[0], 'address']} y su superficie es de {ds.loc[casa_mas_pequenha.index[0], 'surface']}")
```

#### Ejercicio 07. ¿Cuál es la media de precios en la población de "Arroyomolinos (Madrid)"? (★★☆)

#### Ejercicio 08. Trazar el histograma de los precios para la población de "Arroyomolinos (Madrid)" y explica qué observas (★★☆)
